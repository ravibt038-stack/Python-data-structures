## Why python built-in data structure

To collect data in a structured manner

If you collecting more than one value then the data needs to be stored in a data structure






### Types of data structures

1. List
2. Sets
3. Tuples
4. Dictionary - (Key - Value)



# 1.List

1. List accepts duplicate value
2. One dimensional
3. Heterogenous data structure (collecting multiple datatypes in same list)
4. Supports Indexing 
    * positive indexing starts from 0 and negative index starts from -1 
    * Always use square bracket for indexing (regardless of the datastructure)
5. Supports slicing
    * Slicing needs to be provided with range using colon in square bracket
    * [:] left side of colon indicates start position and right side stop position
    * Right part of the colon will always print one value less
6. Mutable - Can be changed anytime in future
7. Create empty list by list=[]

```python
list=["a","b","c","a","a"] #to initiate list use square bracket
```

```python
list
```

```python
list[3] #indexing
```

```python
list[0:5]
```

```python
list[-4:-1] #slicing
```

```python
list[1]="z" #Mutable
```

```python
list
```

6. Mutable - Can be deleted, removed or added but wont work like list
7. Empty set Set=set() and value added set set_list={}
8. Takes item in a order first float,int, alpahbetical order

# 1.Tuple

1. Tuple accepts duplicate value
2. One dimensional
3. Heterogenous data structure (collecting multiple datatypes)
4. Supports Indexing 
5. Supports slicing
6. Immutable - cannot be changed
7. Empty tuple Tuple=() or Tuple=20,60,78,95}
8. Not sorted, retains the value as entered

```python
empty_tuple=()
```

```python
items_tuple="Apple","Kiwi","Bannana",1.2,1000,60 #parantheses also can be used
```

```python
type(items_tuple)
```

```python
items_tuple[2] #indexing
```

```python
items_tuple[2:5] #slicing
```

```python
items_tuple.count(1.2) #functions in tuple
```

```python
items_tuple[2]="Cashwe" #immutable
```

## 1.Dictionary :(Key:Value pairs)
1. Dictionary key doesnot accept duplicates while the values can be duplicated
2. Two dimensional
3. Heterogenous data structure (collecting multiple datatypes)
4. Supports Indexing by calling the key in square brackets
5. Not support slicing
6. Mutable
7. Empty Dictonary Dictionary={} or Dictionary={"Ravi":20,"Shiva":60,"Kamat": 78,"Udupi":95}
8. Not sorted, retains the value as entered
9. Can be used in libraries like panda for better representation

```python
Dictionary={"Ravi":20,"Shiva":60,"Kamat": 78,"Udupi":95}
```

```python
Dictionary
```

```python
Dictionary_list={"Ravi":[20,30,40],"Shiva":[20,30,40],"Kamat":[20,30,40],"Udupi":[20,30,40]}
```

```python
Dictionary_list
```

```python
import pandas as pd
pd.DataFrame(data=Dictionary) # It treats key as columns and values as rows. Rows should have more than one value to form a data frame
```

```python
import pandas as pd
pd.DataFrame(data=Dictionary_list) # Rows(Values) should have more than one instance to form a data frame
```
